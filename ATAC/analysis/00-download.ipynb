# Download

The SC-Framework provides all files used in the snATAC analysis through a convenient download function. All files will be downloaded to `../data/ATAC/`, which is the expected input path for the following notebooks.

The snATAC analysis utilizes the ATAC [10k human PBMC dataset](https://www.10xgenomics.com/datasets/10k-human-pbmcs-atac-v2-chromium-controller-2-standard) from 10X Genomics. Auxiliary files are provided to enable steps such as feature annotation.
#### 10X Genomics
- filtered_peak_bc_matrix
- 10k_pbmc_ATACv2_nextgem_Chromium_Controller_fragments.tsv
- 10k_pbmc_ATACv2_nextgem_Chromium_Controller_possorted_bam.bam

#### Axiliary files
- hg38.blacklist.v2.bed ([ENCODE blacklist](https://doi.org/10.5281/zenodo.1491733))
- JASPAR2024_CORE_vertebrates_non-redundant_pfms_jaspar.txt ([JASPAR](https://doi.org/10.1093/nar/gkad1059))
- homo_sapiens.110.mainChr.fa ([ENSEMBL](https://doi.org/10.1093/nar/gkac958) release 110)
- homo_sapiens.110.genes.gtf ([ENSEMBL](https://doi.org/10.1093/nar/gkac958) release 110)
- homo_sapiens.110.promoters2000.gtf (corresponds to the genes of the above file. Assigns the 2000bp upstream of each gene as promoter.)

The total size after unpacking is **~52GB**.
